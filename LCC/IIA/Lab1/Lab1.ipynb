````python
from abc import ABC, abstractmethod
from typing import Generic, TypeVar
from typing import Any, Optional
from typing import List, Tuple
````

Vamos a crear una clase abstracta ProblemaEE que define el contrato (interfaz) que deben implementar todos los problemas concretos. Luego, los algoritmos de búsqueda trabajarán sobre cualquier instancia que siga este contrato.

````python
# Definimos un tipo genérico para el estado
EstadoEE = TypeVar('EstadoEE')

class ProblemaEE(ABC, Generic[EstadoEE]):
    """
    Clase base abstracta para representar un problema de espacio de estados.
    """
    @abstractmethod
    def estado_inicial(self) -> EstadoEE:
        """Devuelve el estado inicial del problema."""
        pass

    @abstractmethod
    def es_estado_meta(self, estado: EstadoEE) -> bool:
        """Verifica si un estado dado es un estado meta."""
        pass

    @abstractmethod
    def sucesores(self, estado: EstadoEE) -> List[Tuple[EstadoEE, float]]:
        """
        Devuelve una lista de tuplas (estado_siguiente, costo) alcanzables desde el estado dado.
        Si el problema no tiene costos variables, puede usarse 1 como costo por defecto.
        """
        pass

    def heuristica(self, estado: EstadoEE) -> float:
        """
        Heurística para el estado dado (opcional). Por defecto devuelve 0.
        Solo se usa en algoritmos informados como A*.
        """
        return 0.0
````

## BFS

````python
def bfs(problema: ProblemaEE[EstadoEE]) -> Optional[List[EstadoEE]]:
    estado_inicial = problema.estado_inicial()

    # La frontera guarda tuplas de (estado_actual, camino_recorrido)
    frontera = deque([(estado_inicial, [])])
    visitados = set()

    while frontera:
        # Extraemos el primer elemento (FIFO)
        estado, camino = frontera.popleft()
        if problema.es_estado_meta(estado):
            return camino + [estado]

        if estado in visitados:
            continue
        visitados.add(estado)

        for sucesor, _ in problema.sucesores(estado):
            if sucesor not in visitados:
                # En BFS agregamos al final de la cola
                frontera.append((sucesor, camino + [estado]))

    return None # No se encontró solución
````

````python
def bfs_con_traza(problema: ProblemaEE[EstadoEE]):
    """
    Versión de BFS que imprime la tabla de exploración paso a paso.
    """
    estado_inicial = problema.estado_inicial()
    frontera = deque([(estado_inicial, [])])
    visitados = set()
    paso = 1

    print(f"{'Paso':<5} | {'Nodo Actual':<25} | {'Camino':<35} | {'Frontera (Próximos)':<30}")
    print("-" * 100)

    while frontera:
        estado_actual, camino = frontera.popleft()

        # Representación visual de la frontera para la tabla
        nodos_en_frontera = [str(est) for est, cam in frontera]
        str_frontera = ", ".join(nodos_en_frontera) if nodos_en_frontera else "Vacia"

        # Imprimimos la fila de la tabla
        print(f"{paso:<5} | {str(estado_actual):<25} | {str(camino):<35} | {str_frontera:<30}")

        if problema.es_estado_meta(estado_actual):
            print("-" * 100)
            print(f"¡Meta encontrada en el paso {paso}!")
            return camino + [estado_actual]

        if estado_actual not in visitados:
            visitados.add(estado_actual)

            for sucesor, costo in problema.sucesores(estado_actual):
                if sucesor not in visitados:
                    frontera.append((sucesor, camino + [estado_actual]))

        paso += 1

    print("No se encontró solución.")
    return None
````
